Nhập nhiệt độ theo Celsius, in ra nhiệt độ tương ứng theo Fahrenheit.

**Công thức:** `F = C × 9/5 + 32`

```

```

````python
c = float(input("Nhập nhiệt độ (°C): "))

# TODO: Áp dụng công thức đổi sang Fahrenheit
f = c*9/5 +32

print(f"{c} °C = {f} °F")
````

## Bài 3. Ghép họ tên và định dạng chuỗi

Nhập riêng **họ**, **tên đệm**, **tên** rồi in ra họ tên đầy đủ theo định dạng chuẩn (title case) và đếm số ký tự (không kể khoảng trắng).

```
Nhập họ      : nguyễn
Nhập tên đệm : văn
Nhập tên     : an

Họ tên đầy đủ : Nguyễn Văn An
Số ký tự      : 10
Viết hoa      : NGUYỄN VĂN AN
```

````python
ho       = input("Nhập họ      : ").strip()
ten_dem  = input("Nhập tên đệm : ").strip()
ten      = input("Nhập tên     : ").strip()

# TODO: Ghép thành họ tên đầy đủ, chuyển về title case
ho_ten = f'{ho.title()} {ten_dem.title()} {ten.title()}'

# TODO: Đếm ký tự không kể khoảng trắng
so_ky_tu = len(ho_ten.replace(" ",""))

print()
print("Họ tên đầy đủ :", ho_ten)
print("Số ký tự      :", so_ky_tu)
print("Viết hoa      :", ho_ten.upper())
````

## Bài 4. Tính tiền sau thuế VAT

Nhập giá gốc sản phẩm (VNĐ) và thuế suất VAT (%). Tính và in ra tiền thuế phải trả và tổng tiền thanh toán.

```
Giá gốc (VNĐ)  : 500000
```

## Bài 5. Tính chỉ số BMI

Nhập **cân nặng** (kg) và **chiều cao** (cm). Tính BMI và in ra phân loại.

**Công thức:** `BMI = cân_nặng / (chiều_cao_mét ** 2)`

| BMI | Phân loại |
|-----|-----------|
| < 18.5 | Thiếu cân |
| 18.5 – 24.9 | Bình thường |
| 25.0 – 29.9 | Thừa cân |
| ≥ 30.0 | Béo phì |

```
Cân nặng (kg) : 70
Chiều cao (cm): 175

BMI của bạn   : 22.86
Phân loại     : Bình thường
```

````python
can_nang   = float(input("Cân nặng (kg) : "))
chieu_cao  = float(input("Chiều cao (cm): "))

# TODO: Đổi chiều cao từ cm sang mét
chieu_cao_met = chieu_cao/100

# TODO: Tính BMI
bmi = can_nang/(chieu_cao_met**2)

# TODO: Phân loại
if bmi < 18.5:
    phan_loai = "Thiếu cân"
elif bmi<24.9:
    phan_loai = "Bình thường"
elif bmi<29.9:
    phan_loai = "Thừa cân"
else:
    phan_loai = "Béo phì"

print()
print(f"BMI của bạn   : {bmi:.2f}")
print(f"Phân loại     : {phan_loai}")
````

## Bài 6. Đổi giây → giờ, phút, giây

Nhập vào số giây (số nguyên). In ra dạng `giờ:phút:giây`.

**Gợi ý:** Dùng phép chia nguyên `//` và chia lấy dư `%`.

```
Nhập số giây: 3661
3661 giây = 1 giờ 1 phút 1 giây

Nhập số giây: 7384
7384 giây = 2 giờ 3 phút 4 giây
```

````python
tong_giay = int(input("Nhập số giây: "))

# TODO: Tính số giờ, phút, giây còn lại
gio   = tong_giay//3600
phut  = (tong_giay%3600)//60
giay  = (tong_giay%3600)%60

print(f"{tong_giay} giây = {gio} giờ {phut} phút {giay} giây")
````

## Bài 7. Tính tiền lẻ (chia tiền)

Nhập số tiền (VNĐ) cần trả. Tính xem cần bao nhiêu tờ tiền mệnh giá **500.000, 200.000, 100.000, 50.000, 20.000, 10.000**.

*(Ưu tiên tờ to nhất trước, lấy ít tờ nhất có thể)*

```
Số tiền cần trả (VNĐ): 1380000

500,000 VNĐ : 2 tờ
200,000 VNĐ : 1 tờ
100,000 VNĐ : 1 tờ
 50,000 VNĐ : 1 tờ
 20,000 VNĐ : 1 tờ
 10,000 VNĐ : 1 tờ
```

````python
so_tien = int(input("Số tiền cần trả (VNĐ): "))

menh_gia = [500_000, 200_000, 100_000, 50_000, 20_000, 10_000]

print()
con_lai = so_tien
for to in menh_gia:
    # TODO: Tính số tờ tiền mệnh giá 'to' cần dùng
    so_to  = con_lai//to
    con_lai -= so_to*to
    if so_to > 0:
        print(f"{to:>10,} VNĐ : {so_to} tờ")

if con_lai > 0:
    print(f"(Không đủ mệnh giá cho {con_lai:,} VNĐ còn lại)")
````

## Bài 8. Chu vi và diện tích hình tròn

Nhập bán kính `r`. In ra **chu vi** và **diện tích** hình tròn, làm tròn 4 chữ số thập phân.

**Công thức:** `C = 2πr` &nbsp;&nbsp; `S = πr²`

```
Nhập bán kính: 7

Chu vi   : 43.9823
Diện tích: 153.9380
```

**Gợi ý:** Dùng `import math` và `math.pi`

````python
import math

r = float(input("Nhập bán kính: "))

# TODO: Tính chu vi và diện tích
chu_vi    = 2*math.pi*r
dien_tich = math.pi*(r**2)

print()
print(f"Chu vi   : {chu_vi:.4f}")
print(f"Diện tích: {dien_tich:.4f}")
````

## Bài 10. Tính điểm trung bình và xếp loại

Nhập điểm 3 môn học (thang điểm 10, có thể là số thực). Tính điểm trung bình và xếp loại.

| Điểm TB | Xếp loại |
|---------|----------|
| ≥ 9.0 | Xuất sắc |
| ≥ 8.0 | Giỏi |
| ≥ 7.0 | Khá |
| ≥ 5.0 | Trung bình |
| < 5.0 | Yếu |

```
Điểm môn 1: 8.5
Điểm môn 2: 7.0
Điểm môn 3: 9.0

Điểm trung bình: 8.17
Xếp loại       : Giỏi
```

````python
diem1 = float(input("Điểm môn 1: "))
diem2 = float(input("Điểm môn 2: "))
diem3 = float(input("Điểm môn 3: "))

# TODO: Tính điểm trung bình
diem_tb = (diem1+diem2+diem3)/3

# TODO: Xếp loại theo bảng trên
if diem_tb>=9:
    xep_loai = "Xuất sắc"
elif diem_tb>=8:
    xep_loai = "Giỏi"
elif diem_tb>=7:
    xep_loai = "Khá"
elif diem_tb>=5:
    xep_loai = "Trung bình"
else:
    xep_loai = "Yếu"

print()
print(f"Điểm trung bình: {diem_tb:.2f}")
print(f"Xếp loại       : {xep_loai}")
````
